```python
# Imports
import numpy as np
import matplotlib.pyplot as plt
from dataclasses import dataclass

np.random.seed(42)  # reproducibility

# Small helper to add a bias (column of 1s)
def add_bias(X):
    return np.c_[np.ones((X.shape[0], 1)), X]

# Prediction for 1D (w = [bias, slope])
def predict_1d(x, w):
    Xb = add_bias(x.reshape(-1, 1))
    return Xb @ w

# Mean Squared Error
def mse(y_true, y_pred):
    e = y_pred - y_true
    return np.mean(e**2)

# Plot scatter + a line y = w0 + w1 x
def plot_scatter_and_line(x, y, w=None, title="", legend_line_label="Fit"):
    plt.figure()
    plt.scatter(x, y, alpha=0.7)
    if w is not None:
        xs = np.linspace(x.min(), x.max(), 100)
        ys = predict_1d(xs, w)
        plt.plot(xs, ys, label=legend_line_label)
        plt.legend()
    plt.title(title)
    plt.xlabel("x (feature)")
    plt.ylabel("y (target)")
    plt.show()

# Plot residual sticks: vertical lines from each point to the fitted line
def plot_residuals(x, y, w, title="Residuals (vertical distances)"):
    plt.figure()
    plt.scatter(x, y, alpha=0.6, label="Data")
    xs = np.linspace(x.min(), x.max(), 100)
    ys = predict_1d(xs, w)
    plt.plot(xs, ys, label="Fitted line")
    # vertical sticks
    yhat = predict_1d(x, w)
    for xi, yi, yhi in zip(x, y, yhat):
        plt.plot([xi, xi], [yi, yhi], linewidth=1)
    plt.title(title)
    plt.xlabel("x")
    plt.ylabel("y")
    plt.legend()
    plt.show()

@dataclass
class GDHistory:
    weights: list
    losses: list
```

## 1) Build a simple dataset (1 feature → 1 line)
    Story: “More floor area → higher house price”, with a bit of noise.

```python
# Synthetic 1D data: y = 3.5x + 20 + noise
n = 120
x = np.linspace(20, 120, n)
true_w = np.array([20.0, 3.5])  # [bias, slope]
noise = np.random.normal(0, 15, size=n)
y = predict_1d(x, true_w) + noise

plot_scatter_and_line(x, y, None, title="Step 1: Raw data (scatter)")
```

## 2) Fit Linear Regression — Closed Form (solve all at once)
    We find the “best” line in one shot using normal equations.

```python
# Normal equations: w = (X^T X)^(-1) X^T y
Xb = add_bias(x.reshape(-1, 1))
w_closed = np.linalg.pinv(Xb.T @ Xb) @ (Xb.T @ y)

print("Closed-form weights [bias, slope]:", w_closed)
plot_scatter_and_line(x, y, w_closed, title="Step 2: Closed-form fit", legend_line_label="Closed-form line")
```

## 3) See the errors (residuals)
    The model tries to make these vertical distances small (on average).

```python
plot_residuals(x, y, w_closed, title="Step 3: Residuals to the fitted line")
```

## 4) Fit Linear Regression — Gradient Descent (learn by nudging)
    We start with a guess for the line and keep nudging it to reduce MSE.

```python
def gradient_descent_linear_regression_1d(x, y, lr=0.0005, steps=500):
    Xb = add_bias(x.reshape(-1, 1))
    w = np.zeros(2)  # [bias, slope]
    history = GDHistory(weights=[w.copy()], losses=[mse(y, Xb @ w)])
    n = len(y)

    for _ in range(steps):
        yhat = Xb @ w
        grad = (2/n) * (Xb.T @ (yhat - y))  # gradient of MSE wrt w
        w = w - lr * grad
        history.weights.append(w.copy())
        history.losses.append(mse(y, Xb @ w))
    return w, history

w_gd, hist = gradient_descent_linear_regression_1d(x, y, lr=0.0005, steps=600)
print("GD weights [bias, slope]:", w_gd)

# Loss curve
plt.figure()
plt.plot(hist.losses)
plt.title("Step 4: Gradient Descent — Loss vs Iterations")
plt.xlabel("Iteration")
plt.ylabel("MSE")
plt.show()

# Show line snapshots at different moments
def plot_gd_snapshots(x, y, hist, snapshots=(0, 1, 5, 20, 200, -1)):
    plt.figure()
    plt.scatter(x, y, alpha=0.6, label="Data")
    xs = np.linspace(x.min(), x.max(), 100)
    for s in snapshots:
        w = hist.weights[s]
        ys = predict_1d(xs, w)
        label = f"iter {s if s>=0 else len(hist.weights)-1}"
        plt.plot(xs, ys, label=label)
    plt.title("Step 4: GD line moving towards the best fit")
    plt.xlabel("x")
    plt.ylabel("y")
    plt.legend()
    plt.show()

plot_gd_snapshots(x, y, hist)
```
